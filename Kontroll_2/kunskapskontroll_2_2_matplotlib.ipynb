### Explain shortly what a figure, axes, axis and an artist is in Matplotlib.

```python
# The figure is the object that holds the data that the diagram, axes,  contains.  

# axes is the diagram itself, can be more than one and is in the figure. A diagram have its own values and style to display.
#   It contains the plotted data, axis ticks, labels, title etc.

# axis is usually X and Y but can be XYZ if 3D diagram . A dimension in a diagram with a scale of ticks to show datapoints according to.

# artist is to customize colors, linewitdh, linestyle etc on the diagram.
```

### Create a plot of the function y = x^2 [from -4 to 4, hint use the np.linspace function] both in the object-oriented approach and the pyplot approach. Your plot should have a title and axis-labels.

```python
value_array = np.linspace(-4, 4, 20)# 20 numbers between -4 - 4

# OO style
fig, ax = plt.subplots(figsize=(5, 2.7))

ax.plot(value_array, value_array**2, label='Style as OO') 
ax.set_xlabel('x values OO')  
ax.set_ylabel('y values OO')  
ax.set_title("Graph as OO")  



# pyplot style
plt.figure(figsize=(5, 2.7))
plt.plot(value_array, value_array**2, label='Style as pyplot') 
plt.xlabel('x values pyplot')
plt.ylabel('y values pyplot')
plt.title("Graph as pyplot")

# plt.show() # not needed in jupyter
```

### Create a figure containing 2  subplots where the first is a scatter plot and the second is a bar plot. You have the data below. 

```python
# Data for scatter plot
np.random.seed(15)
random_data_x = np.random.randn(1000)
random_data_y = np.random.randn(1000)

# Data for bar plot
fruit_data = {'grapes': 22, 'apple': 8, 'orange': 15, 'lemon': 20, 'lime': 25}
names = list(fruit_data.keys())
values = list(fruit_data.values())
```

```python
fig, axs = plt.subplots(1, 2, figsize=(15, 6)) # 2 diagrams

axs[0].scatter(random_data_x, random_data_y)
axs[1].bar(x=names, height=values) 
```
